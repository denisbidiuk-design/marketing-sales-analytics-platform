## #️⃣ 1. Import Libraries

```python
import pandas as pd
import numpy as np
import datetime
```

## #️⃣ 2. Cleaning Contacts

### 📥 Loading Data

```python
contacts_df = pd.read_excel("Contacts (CRM).xlsx")
contacts_df.head()
contacts_df.info()
```

- non‑optimal data types

- unnecessary columns

- missing values

✔ Dates in datetime

✔ Categorical fields cleaned

✔ Missing values handled

```python
contacts_df.to_excel("Contacts (Result).xlsx", index=False)
```

## #️⃣ 3. Cleaning Calls

### 📥 Loading Data

```python
calls_df = pd.read_excel("Calls (CRM).xlsx")
calls_df.head()
calls_df.info()
```

### 🧩 Issues Before Cleaning
- completely empty columns (Dialled Number, Tag)

- missing values in Outgoing Call Status and Scheduled in CRM

- dates stored as text

- non‑optimal data types

### 🗑 Removing Empty Columns

```python
cols_to_drop = ["Dialled Number", "Tag"]
existing_cols = [c for c in cols_to_drop if c in calls_df.columns]

calls_df.drop(columns=existing_cols, inplace=True)
```

### 🔧 Converting Call Date

```python
calls_df["Call Start Time"] = pd.to_datetime(
    calls_df["Call Start Time"],
    format="%d.%m.%Y %H:%M",
    dayfirst=True,
    errors="coerce"
)
```

### 🏷 Converting Categorical Columns

```python
category_cols = [
    "Call Owner Name", "Call Type", "Call Status", "Outgoing Call Status"
]

for col in category_cols:
    calls_df[col] = calls_df[col].astype("category")
```

### 🩹 Handling Missing Values

```python
# Outgoing Call Status
calls_df["Outgoing Call Status"] = calls_df["Outgoing Call Status"].astype("category")
calls_df["Outgoing Call Status"] = calls_df["Outgoing Call Status"].fillna("Unknown")

# Call Duration
calls_df["Call Duration (in seconds)"] = (
    calls_df["Call Duration (in seconds)"]
    .fillna(0)
    .astype(int)
)

# Scheduled in CRM
calls_df["Scheduled in CRM"] = (
    calls_df["Scheduled in CRM"]
    .fillna(False)
    .astype(bool)
)
```

### 📊 Final Structure

```python
calls_df.info()
calls_df.isna().sum()
```

✔ Dates converted to datetime  

✔ Categorical fields optimized

✔ Empty columns removed

✔ Missing values handled

✔ Memory usage reduced by ~2.5×

### 💾 Saving Results

```python
calls_df.to_excel("Calls (Result).xlsx", index=False)
```

## #️⃣ 4. Cleaning **Spend**

## #️⃣ 5. Cleaning Deals

### 📥 Loading Data

```python
deals_df = pd.read_excel("Deals (CRM).xlsx")
deals_df.head()
deals_df.info()
```

### 🧩 Issues Before Cleaning
- currency fields stored as text

- SLA stored as time or timedelta

- missing values in Contact Name

- swapped values in Initial Amount Paid and Offer Total Amount

- incorrect Closing Date

- duplicates

### 💰 Cleaning Currency Columns

```python
def clean_currency(df, col):
    s = df[col].astype(str)
    s = s.str.replace("€", "", regex=False)
    s = s.str.replace(" ", "", regex=False)
    s = s.apply(lambda x: x if x.count(".") <= 1 else x.replace(".", "", x.count(".") - 1))
    s = s.str.replace(",", ".", regex=False)
    s = s.replace("", np.nan)
    df[col] = pd.to_numeric(s, errors="coerce")
    return df

deals_df = clean_currency(deals_df, "Initial Amount Paid")
deals_df = clean_currency(deals_df, "Offer Total Amount")
```

### ⏱ Converting SLA → Seconds

```python
def convert_to_seconds(x):
    if pd.isna(x):
        return np.nan
    if isinstance(x, datetime.time):
        return x.hour * 3600 + x.minute * 60 + x.second
    if isinstance(x, datetime.timedelta):
        return x.total_seconds()
    return np.nan

if "SLA" in deals_df.columns:
    deals_df["SLA Seconds"] = deals_df["SLA"].apply(convert_to_seconds)
    deals_df.drop(columns=["SLA"], inplace=True)
```

### 📅 Converting Date Columns

```python
deals_df["Closing Date"] = pd.to_datetime(
    deals_df["Closing Date"],
    format="%d.%m.%Y",
    dayfirst=True,
    errors="coerce"
)

deals_df["Created Time"] = pd.to_datetime(
    deals_df["Created Time"],
    format="%d.%m.%Y %H:%M",
    dayfirst=True,
    errors="coerce"
)

deals_df["Created Date"] = pd.to_datetime(deals_df["Created Time"].dt.date)
```

### 🩹 Handling Missing Values

```python
deals_df["Stage"] = deals_df["Stage"].astype(str).str.strip()

# Product normalization
valid_products = ["Digital Marketing", "UX/UI Design", "Web Developer"]
deals_df["Product"] = deals_df["Product"].astype(str).str.strip()
deals_df.loc[~deals_df["Product"].isin(valid_products), "Product"] = "Other"
```

### 🔄 Fixing Swapped Payment Columns

```python
mask_swap = deals_df["Initial Amount Paid"] > deals_df["Offer Total Amount"]
deals_df.loc[mask_swap, ["Initial Amount Paid", "Offer Total Amount"]] = (
    deals_df.loc[mask_swap, ["Offer Total Amount", "Initial Amount Paid"]].values
)
```

### 💳 Restoring Payment Type

```python
deals_df.loc[
    deals_df["Initial Amount Paid"] == deals_df["Offer Total Amount"],
    "Payment Type"
] = "One Payment"

deals_df.loc[
    deals_df["Initial Amount Paid"] < deals_df["Offer Total Amount"],
    "Payment Type"
] = "Partial Payment"
```

### 🧹 Filling Numeric Fields

```python
deals_df["Initial Amount Paid"] = pd.to_numeric(deals_df["Initial Amount Paid"], errors="coerce")
deals_df["Offer Total Amount"] = pd.to_numeric(deals_df["Offer Total Amount"], errors="coerce")
deals_df["Months of study"] = pd.to_numeric(deals_df["Months of study"], errors="coerce")
deals_df["Course duration"] = pd.to_numeric(deals_df["Course duration"], errors="coerce")
```

### 📊 Final Structure

```python
deals_df.info()
```

### 💾 Saving Results

```python
deals_df.to_excel("Deals (Result).xlsx", index=False)
```
